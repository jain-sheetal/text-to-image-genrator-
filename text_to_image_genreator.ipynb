## 🎯 Module 2: Conditional GAN (CGAN) for Categorical Visuals
In Conditional GANs, both the Generator (z, c)$ and Discriminator (x, c)$ receive condition labels (e.g., shapes: circle, square, triangle).

```python
from src.cgan import CGANPipeline, SHAPE_CLASSES
cgan = CGANPipeline(device="cuda" if torch.cuda.is_available() else "cpu")
print(f"Available CGAN Shape Classes: {SHAPE_CLASSES}")
shape_img, meta = cgan.generate("circle", seed=42)
print("CGAN Diagnostics:", meta)
display(shape_img)
```

## 🔤 Module 3: Hugging Face Transformers Text Preprocessing & Embeddings
Tokenize prompt descriptions into token IDs and continuous 512-dimensional semantic embeddings.

```python
from src.text_processor import TextPreprocessor
processor = TextPreprocessor()
prompt = "a vibrant red flower blooming in sunshine with dew drops"
token_info = processor.tokenize(prompt)
print(f"Active Tokens: {token_info['active_tokens']}")
emb, stats = processor.encode_to_embedding(prompt)
print("Embedding Vector Statistics:", stats)
display(processor.generate_token_diagnostic_plot(token_info))
```

## 👁️ Module 5: Attention-Enhanced GAN (Self-Attention & Cross-Attention)
